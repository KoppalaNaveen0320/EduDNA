# EduDNA - Exploratory Data Analysis (EDA)
This notebook analyzes student engagement metrics, quiz performance, and attendance records to understand learning behavior patterns.

```python
import pandas as pd
import numpy as np

# Load datasets
students_df = pd.read_csv('../backend/datasets/synthetic_students.csv')
attendance_df = pd.read_csv('../backend/datasets/attendance.csv')
quiz_df = pd.read_csv('../backend/datasets/quiz_data.csv')

print("Students Dataset Shape:", students_df.shape)
students_df.head()
```

## 1. Learning Modality Distribution
Breakdown of student cognitive learning styles across the dataset.

```python
modality_counts = students_df['learning_style'].value_counts()
print("Modality Distribution:\n", modality_counts)

mean_mastery_by_style = students_df.groupby('learning_style')['mastery_rate'].mean()
print("\nAverage Mastery Rate by Modality:\n", mean_mastery_by_style)
```

## 2. Quiz Performance & Concept Mastery
Examine quiz score variation across concepts.

```python
topic_performance = quiz_df.groupby('topic').agg({
    'score': ['mean', 'min', 'max'],
    'attempts': 'mean',
    'time_spent_secs': 'mean'
})
print("Performance by Topic:\n", topic_performance)
```
