# EduDNA - Learning Genome & GNN Model Training
This notebook trains Graph Neural Network (GNN) embeddings to map concept mastery and generate learning genome vectors.

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import pandas as pd

print("PyTorch Version:", torch.__version__)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print("Using device:", device)
```

## 1. GNN Knowledge Graph Architecture
Define a 2-layer Graph Convolutional Network (GCN) for predicting concept mastery transitions.

```python
class ConceptGNN(nn.Module):
    def __init__(self, in_features, hidden_dim, out_features):
        super(ConceptGNN, self).__init__()
        self.fc1 = nn.Linear(in_features, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, out_features)
        self.dropout = nn.Dropout(0.2)

    def forward(self, x, adj):
        # Graph convolution operation
        x = torch.spmm(adj, x)
        x = F.relu(self.fc1(x))
        x = self.dropout(x)
        x = torch.spmm(adj, x)
        x = torch.sigmoid(self.fc2(x))
        return x

model = ConceptGNN(in_features=8, hidden_dim=16, out_features=4).to(device)
print(model)
```
